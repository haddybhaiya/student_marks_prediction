```python
# label encoding for the columns
le = LabelEncoder()
dataset['gender'] = le.fit_transform(dataset['gender'])
dataset['course'] = le.fit_transform(dataset['course'])
dataset['internet_access'] = le.fit_transform(dataset['internet_access'])
dataset['study_method'] = le.fit_transform(dataset['study_method'])
dataset['sleep_quality'] = le.fit_transform(dataset['sleep_quality'])

#ordinal encoding for facility_rating and exam_difficulty
dataset['facility_rating'] = dataset['facility_rating'].map({ 'high': 3, 'medium': 2, 'low': 1})
dataset['exam_difficulty'] = dataset['exam_difficulty'].map({ 'hard': 3, 'moderate': 2, 'easy': 1})

```

```python
test_dataset = pd.read_csv('test.csv')
test_ids = test_dataset['id']
test_dataset.drop(['id'], axis=1, inplace=True)

```

```python
# Define the unique categories for each column based on the training data's original values,
# sorted alphabetically, which is how LabelEncoder assigns integers.

gender_categories = ['female', 'male', 'other']
course_categories = ['ba', 'b.com', 'b.sc', 'b.tech', 'bba', 'bca', 'diploma']
internet_access_categories = ['no', 'yes']
sleep_quality_categories = ['average', 'good', 'poor']
study_method_categories = ['coaching', 'group study', 'mixed', 'online videos', 'self-study']

test_dataset = pd.read_csv('test.csv')
test_ids = test_dataset['id']
test_dataset.drop(['id'], axis=1, inplace=True)
```

```python
# Handle potential numerical strings in categorical columns by mapping them back to original strings
# For 'gender', map '0'->'female', '1'->'male', '2'->'other' based on LabelEncoder's alphabetical mapping
def map_gender_values(value):
    if value == '0': return 'female'
    if value == '1': return 'male'
    if value == '2': return 'other'
    return value
test_dataset['gender'] = test_dataset['gender'].astype(str).apply(map_gender_values)
```

```python
# Instantiate and fit LabelEncoders for each column with their respective categories
le_gender = LabelEncoder()
le_gender.fit(gender_categories)

le_course = LabelEncoder()
le_course.fit(course_categories)

le_internet_access = LabelEncoder()
le_internet_access.fit(internet_access_categories)

le_study_method = LabelEncoder()
le_study_method.fit(study_method_categories)

le_sleep_quality = LabelEncoder()
le_sleep_quality.fit(sleep_quality_categories)
```

```python
#  Transform the test_dataset columns using the fitted LabelEncoders
test_dataset['gender'] = le_gender.transform(test_dataset['gender'].astype(str))
test_dataset['course'] = le_course.transform(test_dataset['course'].astype(str))
test_dataset['internet_access'] = le_internet_access.transform(test_dataset['internet_access'].astype(str))
test_dataset['study_method'] = le_study_method.transform(test_dataset['study_method'].astype(str))
test_dataset['sleep_quality'] = le_sleep_quality.transform(test_dataset['sleep_quality'].astype(str))
```

```python
# Apply ordinal encoding for facility_rating and exam_difficulty
test_dataset['facility_rating'] = test_dataset['facility_rating'].astype(str).map({ 'high': 3, 'medium': 2, 'low': 1})
test_dataset['exam_difficulty'] = test_dataset['exam_difficulty'].astype(str).map({ 'hard': 3, 'moderate': 2, 'easy': 1})

# Create engineered features
test_dataset['study_efficiency'] = test_dataset['study_hours'] * test_dataset['class_attendance']
test_dataset['sleep_study_ratio'] = test_dataset['sleep_hours'] / (test_dataset['study_hours'] + 1)
```
